```python
# Tuple Methods 
print(numbers.count(1)) # Count function to count how many times it is present. 
print(numbers.index(5))
```

```python
# Packing & Unpacking Tuple 
#Packing Tuple
packed_tuple = 1,"Hello", 3.23
print(packed_tuple)
```

```python
# Unpacking Tuple 
a, b, c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# Unpacking with * 

numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle) #The star take all elements utile the last element is remained. 
print(last)

```

```python
nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False))

#Access the elements inside a tuple 
print(nested_tuple[0])
print(nested_tuple[1] [2])
```

```python
# Iterating over nested tuples 
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
