```python
# Create a set 

my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set = set([1,2,3,4,5,6])
print(type(my_set))
```

```python
my_empty_set = set([1,2,3,4,5,6,5,4,6])
print(my_empty_set)
```

```python
# Basic set operations
# Adding and Removing Elements 

my_set.add(7)
my_set.add(7) #ignore the dublicate elements
print(my_set)
```

```python
# Remove the elements form a set.
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(8)
print(my_set)
```

```python
my_set.discard(8) #this also works same but it doesnot give error when there is no element. 
```

```python
# Pop method 
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
# Clear all the elements 

my_set.clear()
print(my_set)
```

```python
# Set Membership Test
my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
# Mathematical Operations 
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# Union Set (Combine all the elements )
union_set = set1.union(set2)
print(union_set)
```

```python
# Intersection (Only the common elements )
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# Difference 
print(set1.difference(set2))
print(set2.difference(set1))
```

```python
# Symmetric Difference (Both unique elemets of both set)
set1.symmetric_difference(set2)
```

```python
# Sets Methods
set1 = {1,2,3,4,5}
set2 = {3,4,5}

#is subset 
print(set1.issubset(set2))

print(set1.issuperset(set2)) 
```

```python
# Counting Unique words in text
text ="in this tutorial we are discussing about sets."
word = text.split()

# Convert list of words to sets to get unique word
unique_words = set(word)
print(unique_words)
print(len(unique_words))
```
